# Loan Approval Prediction Using Machine Learning

**Internship Company:** Alfido Tech  
**Internship Domain:** Data Science  
**Project:** Loan Approval Prediction  
**Author:** Shubham Khakale  
**Course:** Third Year Engineering – Information Technology (TE IT)

## Project Overview
This project predicts whether a loan application is likely to be approved using machine learning. It uses applicant details such as income, credit history, education, and property area.

## Objectives
- Analyze loan applicant data.
- Clean missing values and duplicate records.
- Visualize important dataset patterns.
- Build and evaluate a classification model.

## Technologies Used
- Python
- Pandas
- Matplotlib
- Seaborn
- Scikit-learn
- JupyterLab

## Dataset
- Total records: 614
- Total columns: 13
- Target column: `Loan_Status`
- Missing values after cleaning: 0

## Methodology
1. Data loading and exploration
2. Data cleaning and preprocessing
3. Exploratory Data Analysis (EDA)
4. Data visualization
5. Random Forest model training
6. Model evaluation

## Model Performance
- **Algorithm:** Random Forest Classifier
- **Accuracy:** 83.74%
- **Test records:** 123
- **Correct predictions:** 103

## Visualizations
The project includes charts for:
1. Loan approval status
2. Applicant income distribution
3. Credit history versus loan approval
4. Property area versus loan approval
5. Confusion matrix
6. Feature importance

## Conclusion
The Random Forest model achieved 83.74% accuracy on the test dataset. It identified approved loans more effectively than loans that were not approved. Further tuning and validation could improve its performance.
